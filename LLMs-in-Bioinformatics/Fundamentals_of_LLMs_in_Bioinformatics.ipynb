## Introduction to LLMs
**Large Language Models (LLMs)** are advanced artificial intelligence systems designed to understand and generate human language. They’re trained on huge amounts of text—like books, articles, and websites—to learn how language works. Built on deep learning techniques, LLMs can perform a wide range of language-based tasks such as answering questions, write summaries, explain complex topics, or even help analyze scientific or medical data.

LLMs are remarkably adaptable. This makes them powerful tools for supporting research, education, and even patient care by making sense of complex information quickly and accurately.

## Prompting
**Prompting** or prompt engineering is the process of guiding a language model by providing specific input that shapes the content and style of its response. A well-crafted prompt acts as a set of instructions, helping the model understand your intent and generate useful, targeted output. In bioinformatics, where precision and domain expertise are critical, the way a prompt is constructed can directly affect the clarity, depth, and relevance of the response. Whether you're generating summaries of genomic data, drafting technical documentation, or exploring complex biological concepts, thoughtful prompt design helps ensure the model responds in a way that aligns with your goals.

---

## **The Primary Components of Effective Prompting**
Crafting a strong prompt is essential to getting accurate, relevant, and well-structured responses from language models. Effective prompts include key components that guide the model's behavior, tone, and focus. By clearly defining elements such as the role, task, and audience, you help the model understand the context and deliver more useful output. The following six primary components—**Role, Task, Length/Format, Tone/Style, Audience, and Restrictions**—work together to shape responses that are coherent, appropriately styled, and aligned with your goals.

To demonstrate how each of the six primary components contributes to an effective prompt, consider the following example. This prompt incorporates all of the key elements to guide the model in producing a focused and contextually appropriate response:

<img src="materials/images/prompt0.png"/>

#### **Role**
Defines the perspective or expertise the model should adopt. This helps shape the response by setting expectations around domain knowledge and approach.

<img src="materials/images/prompt1.png"/>

#### **Task**
Specifies what the model is being asked to do. A well-defined task ensures clarity and relevance in the response.

<img src="materials/images/prompt2.png"/>

#### **Length/Format**
Indicates how long the response should be and/or the desired structure (e.g., paragraphs, bullet points). This guides the output’s scope and organization.

<img src="materials/images/prompt3.png"/>

#### **Tone/Style**
Describes the voice or manner in which the response should be written (e.g., formal, conversational, persuasive). It ensures the response aligns with the context or purpose.

<img src="materials/images/prompt4.png"/>

#### **Audience**
Identifies who the response is intended for. Tailoring the message to the audience ensures appropriate language, depth, and focus.

<img src="materials/images/prompt5.png"/>

#### **Restrictions**
Outlines what to avoid or exclude in the response. These boundaries help maintain relevance and prevent deviation from the prompt's goals.

<img src="materials/images/prompt6.png"/>

While not every prompt will require all six components, keeping them in mind allows you to better tailor and steer the response to meet your specific goals. Depending on the context, some elements—like tone or restrictions—may be more critical than others. The key is to be intentional: the more clearly you define what you need, the more effectively the model can deliver a useful, relevant response.

## **Framing Prompt Engineering as Algorithmic Thinking**
### What is an algorithm?
An algorithm is a step-by-step procedure for solving a problem or performing a task. Similarly, prompt engineering can be seen as designing a process (your prompt) that guides the language model (your tool) through the steps necessary to generate a desired output.

When you write a good prompt, you're effectively giving the model an algorithm to follow—you specify the role, the task, and constraints in a logical, ordered way that leads to a structured response. Just like algorithms need clear inputs and defined processes, prompts need clarity and intentionality.

### Applying Problem-Solving Techniques to Prompt Design
Problem-solving techniques—such as breaking a complex task into subproblems, identifying constraints, or defining the desired outcome—are directly applicable to crafting effective prompts.

- **Working backward:** Start from what you want (a specific type of output) and design the prompt to guide the model toward it.

- **Heuristics:** Use patterns you know work well—like assigning a role or limiting the scope—to get consistent results, just like you’d apply heuristics to solve coding or analysis problems.

- **Decomposition:** Break down a complex information need into smaller parts—task, format, tone, audience. Each of these becomes part of the prompt.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Complex Need:** “Help me understand how variant calling works in exome sequencing data.”

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Decomposed Prompt:**
  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Task:** Provide a clear explanation of the variant calling workflow in exome sequencing.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Format:** Study notes.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Tone:** Educational and clear.

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**Audience:** Bioinformatics trainees or junior data analysts learning core NGS analysis techniques.
>**Prompt:**
>
>Write a set of study notes that clearly explain how variant calling is performed in exome sequencing data.
>
>Break the content into logically organized sections, such as:

In bioinformatics and beyond, we often use LLMs to solve problems—summarizing data, generating code, writing reports. Good prompt engineering ensures the model becomes a useful problem-solving assistant rather than just a generic text generator.

<div class="alert alert-block alert-success">

**Annotating the use of large language models (LLMs)** in bioinformatics—or any scientific context—is a best practice for several important reasons:

1. **Transparency**\
Annotating clearly that a model was used helps readers understand the origin of the content or analysis. It distinguishes human-authored material from AI-generated output, which is especially critical in research and data interpretation where trust and provenance matter.

2. **Reproducibility**\
Science relies on the ability to reproduce results. By documenting the model version and the date of the run, others can attempt to replicate or verify the output under similar conditions—just as they would with software versions or versioned datasets.

3. **Accountability and Ethical Use**\
Clear annotation shows responsible AI use. It signals that the user understands and communicates the role of AI in their workflow, especially important in sensitive fields like healthcare and genomics where AI-generated suggestions should not replace expert judgment.

4. **Acknowledging Limitations**\
LLMs are powerful but not infallible. By stating that an LLM was used, users are implicitly flagging that some content may need further validation. This invites critical scrutiny and avoids the risk of AI-generated errors being mistaken for verified facts.

5. **Academic and Professional Integrity**\
Just as you would cite software tools, databases, or collaborators, crediting the LLM respects intellectual contributions. It’s increasingly expected by journals, institutions, and funding agencies as part of responsible AI integration.

## Sample Use Cases for LLMs in Bioinformatics

#### **Annotating Genomic Variants**
LLMs can help interpret the clinical significance of genetic variants by summarizing research papers, matching variants to known databases, or explaining the impact of a mutation in plain language—useful for both researchers and clinicians.

> **Prompt Example:** _"Explain the functional impact of the BRCA1 c.68_69delAG variant"_ — and the LLM responds with a summary including disease associations and inheritance patterns.


#### <br><br>**Summarizing Research Literature**
LLMs can read and condense complex scientific papers, highlight key findings, or generate quick summaries of large datasets or methods sections—saving time and boosting comprehension.

> **Prompt Example:** _"Summarize the key findings of the 5 most recent research papers on microbiome and cancer from PubMed abstracts."_


#### <br><br>**Natural Language Querying of Datasets**
Instead of writing complex SQL or Python code, researchers can ask questions of the data.

> **Prompt Example:** _"Show me all patients with TP53 mutations and lung cancer"_ — and have the model generate the correct code or retrieve the data.

#### <br><br>**Assisting in Clinical Decision Support**
By integrating with EHRs, LLMs can help summarize patient histories, flag drug-gene interactions, or even suggest follow-up actions based on guidelines or patient-specific data.

> **Prompt Example:** _"Summarize this patient’s last 5 visits and flag any potential drug interactions."_

#### <br><br>**Generating Hypotheses or Study Designs**
LLMs can assist researchers in brainstorming experiments or writing the first draft of a study protocol based on a given research question and dataset.

> **Prompt Example:** _"Suggest an experimental design to study differential gene expression in Alzheimer’s disease using RNA-seq data."_

#### <br><br>**Teaching and Explaining Complex Concepts**
LLMs can act as tutors, breaking down difficult bioinformatics topics like CRISPR, PCA, or GWAS into digestible explanations—tailored for beginners or advanced learners.

> **Prompt Example:** _"Explain PCA in the context of single-cell RNA-seq data to an intermediate biology student."_

#### <br><br>**Explaining Complex Concepts to Non-Experts**
LLMs can translate technical terms (e.g., "haplotype phasing" or "differential gene expression") into layperson-friendly explanations for patient reports or interdisciplinary teams.

> **Prompt Example:** _"Explain variant calling in simple terms suitable for a patient-facing report."_
